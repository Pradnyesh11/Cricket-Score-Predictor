```python
import pickle

with open('t20_score_predictor.pkl','rb')as f:
    model=pickle.load(f)

```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
import pandas as pd
```

```python
def predict_score(city,bowling_team,batting_team,balls_left,wickets_left,crr,last_five,current_score):
    city_enc=le.fit_transform([city])[0]
    batting_team_enc=le.fit_transform([batting_team])[0]
    bowling_team_enc=le.fit_transform([bowling_team])[0]

    input_data=pd.DataFrame({
        'city':[city_enc],
        'bowling_team':[bowling_team_enc],
        'batting_team':[batting_team_enc],
        'balls_left':[balls_left],
        'wickets_left':[wickets_left],
        'crr':[crr],
        'last_five':[last_five],
        'current_score2':[current_score]
    })

    predicted_score=model.predict(input_data)
    return int(predicted_score[0])

    
```

```python
predicted=predict_score(
    city          = 'Colombo',
    bowling_team  = 'Pakistan',
    batting_team  = 'India',
    balls_left    = 10,    
    wickets_left  = 2,     
    crr           = 5.6,   
    last_five     = 20 ,
    current_score = 120
)
print(f"Predicted Final Score: {predicted}")
```
